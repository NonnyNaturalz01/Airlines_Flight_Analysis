<b>Does price varies with airlines</b>

```python
#grouping airlines with their mean prices
data.groupby('airline')['price'].mean()
```

```python
#categorical plot showing the mean ticket price for each airline
sns.catplot(x='airline', y='price', kind= 'bar', palette = 'rocket', data = data, hue='class')
plt.show()
```
